En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

*   `2022 Ranking`: Ranking del año actual de la empresa
*   `Organization Name`: Nombre
*   `Industry`: El tipo de industria con la que trata principalmente

*   `Year Founded`: Año en que se fundó
*   `CEO`: CEO
*   `Profits (Billions)`: Ganancias del año
*   `Assets (Billions)`: Activos del año
*   `Market Value (Billions)`: Valor de mercado del año
*   `Total Employees`: Total de empleados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================================
# CONFIGURACIÓN ESTÉTICA GLOBAL
# =====================================================================
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']
plt.rcParams['axes.edgecolor'] = '#cccccc'
plt.rcParams['axes.linewidth'] = 0.8

# Paleta de colores principal (Tonos corporativos modernos)
PRIMARY_COLOR = '#1f77b4'
ACCENT_PALETTE = 'mako'

# =====================================================================
# 1. MAPA DE CALOR (1 Gráfica)
# =====================================================================
plt.figure(figsize=(10, 8))
numeric_cols = forbes_df.select_dtypes(include=[np.number])
corr = numeric_cols.corr()

# Máscara para la mitad superior opcional (opcional, dejamos completo con mejor colormap)
sns.heatmap(corr, annot=True, cmap='vlag', fmt=".2f", linewidths=.8,
            cbar_kws={'shrink': 0.8}, annot_kws={'size': 10, 'weight': 'bold'})

plt.title('Mapa de Calor: Correlación de Variables Numéricas', fontsize=14, fontweight='bold', pad=15)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# =====================================================================
# 2. DIAGRAMAS DE DISPERSIÓN (3 Gráficas con Línea de Tendencia)
# =====================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

scatter_configs = [
    ('Profits (Billions)', 'Market Value (Billions)', 'Ganancias vs Valor de Mercado', '#2b5c8f'),
    ('Revenue (Billions)', 'Assets (Billions)', 'Ingresos vs Activos', '#2ca02c'),
    ('Total Employees', 'Revenue (Billions)', 'Empleados vs Ingresos', '#d62728')
]

for i, (x_col, y_col, title, color) in enumerate(scatter_configs):
    sns.regplot(data=forbes_df, x=x_col, y=y_col, ax=axes[i],
                color=color, scatter_kws={'alpha': 0.5, 's': 40},
                line_kws={'color': '#333333', 'linewidth': 1.5, 'linestyle': '--'})
    axes[i].set_title(title, fontsize=12, fontweight='bold', pad=10)
    axes[i].set_xlabel(axes[i].get_xlabel(), fontsize=10, fontweight='bold')
    axes[i].set_ylabel(axes[i].get_ylabel(), fontsize=10, fontweight='bold')
    sns.despine(ax=axes[i], top=True, right=True)

plt.tight_layout()
plt.show()

# =====================================================================
# 3. DIAGRAMAS DE CAJA / BOXPLOTS (2 Gráficas con Distribución Real)
# =====================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

box_configs = [
    ('Profits (Billions)', 'Distribución de Ganancias (Billones)', '#5dade2'),
    ('Revenue (Billions)', 'Distribución de Ingresos (Billones)', '#af7ac5')
]

for i, (col, title, color) in enumerate(box_configs):
    # Boxplot principal limpio
    sns.boxplot(data=forbes_df, y=col, ax=axes[i], color=color, fliersize=3, width=0.4, boxprops=dict(alpha=0.8))
    # Superponer stripplot para ver la densidad de los datos
    sns.stripplot(data=forbes_df, y=col, ax=axes[i], color='#333333', alpha=0.2, size=3, jitter=0.2)

    axes[i].set_title(title, fontsize=12, fontweight='bold', pad=10)
    axes[i].set_ylim(-2, forbes_df[col].quantile(0.92))
    axes[i].set_ylabel(col, fontsize=10, fontweight='bold')
    sns.despine(ax=axes[i], top=True, right=True)

plt.tight_layout()
plt.show()

# =====================================================================
# 4. GRÁFICAS DE BARRAS DATOS CATEGÓRICOS (5 Gráficas)
# =====================================================================
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

# Gráfica 1: Top 10 Países
top_countries = forbes_df['Country'].value_counts().head(10)
sns.barplot(x=top_countries.index, y=top_countries.values, ax=axes[0], palette='crest', hue=top_countries.index, legend=False)
axes[0].set_title('Top 10 Países con más Empresas', fontsize=12, fontweight='bold', pad=10)
axes[0].set_ylabel('Cantidad de Empresas', fontsize=10, fontweight='bold')
axes[0].tick_params(axis='x', rotation=45)

# Gráfica 2: Top 10 Industrias
top_industries = forbes_df['Industry'].value_counts().head(10)
sns.barplot(x=top_industries.index, y=top_industries.values, ax=axes[1], palette='crest', hue=top_industries.index, legend=False)
axes[1].set_title('Top 10 Industrias Predominantes', fontsize=12, fontweight='bold', pad=10)
axes[1].set_ylabel('Cantidad', fontsize=10, fontweight='bold')
axes[1].tick_params(axis='x', rotation=45)

# Gráfica 3: Top 10 Años de fundación
if 'Year Founded' in forbes_df.columns:
    top_years = forbes_df[forbes_df['Year Founded'] > 0]['Year Founded'].value_counts().head(10)
    sns.barplot(x=top_years.index.astype(str), y=top_years.values, ax=axes[2], palette='crest', hue=top_years.index, legend=False)
    axes[2].set_title('Top 10 Años de Fundación', fontsize=12, fontweight='bold', pad=10)
    axes[2].set_ylabel('Cantidad', fontsize=10, fontweight='bold')
    axes[2].tick_params(axis='x', rotation=45)

# Gráfica 4: Top 5 Empresas por Ganancias
top_profits = forbes_df.nlargest(5, 'Profits (Billions)')
sns.barplot(data=top_profits, x='Organization Name', y='Profits (Billions)', ax=axes[3], palette='Blues_r', hue='Organization Name', legend=False)
axes[3].set_title('Top 5 Empresas por Ganancias', fontsize=12, fontweight='bold', pad=10)
axes[3].set_ylabel('Ganancias (Billones)', fontsize=10, fontweight='bold')
axes[3].tick_params(axis='x', rotation=30)

# Gráfica 5: Top 5 Empresas por Empleados
top_emp = forbes_df.nlargest(5, 'Total Employees')
sns.barplot(data=top_emp, x='Organization Name', y='Total Employees', ax=axes[4], palette='Greens_r', hue='Organization Name', legend=False)
axes[4].set_title('Top 5 Empresas por Número de Empleados', fontsize=12, fontweight='bold', pad=10)
axes[4].set_ylabel('Total Empleados', fontsize=10, fontweight='bold')
axes[4].tick_params(axis='x', rotation=30)

# Limpiar bordes y ocultar panel sobrante
for i in range(5):
    sns.despine(ax=axes[i], top=True, right=True)
    axes[i].set_xlabel('')

axes[5].axis('off')

plt.tight_layout()
plt.show()
```
