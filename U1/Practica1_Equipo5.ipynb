```python
import pandas as pd
forbes_df = pd.read_csv('Forbes_2000_top_company.csv')
```

```python
import matplotlib.pyplot as plt

#Identificar automáticamente las columnas categóricas (tipo object) del DataFrame
columnas_categoricas = forbes_df.select_dtypes(include=['object']).columns

#Iterar sobre cada columna categórica para generar una gráfica de barras individual
for col in columnas_categoricas:
    plt.figure(figsize=(10, 5))

    #Si la columna tiene muchos valores únicos (como nombres de empresas o CEOs),
    #mostramos únicamente el Top 10 para que la gráfica sea legible y limpia.
    if forbes_df[col].nunique() > 15:
        data_a_graficar = forbes_df[col].value_counts().head(10)
        titulo = f'Top 10 de {col} con mayor frecuencia'
    else:
        data_a_graficar = forbes_df[col].value_counts()
        titulo = f'Distribución de {col}'

    #Crear la gráfica de barras
    data_a_graficar.plot(kind='bar', color='skyblue', edgecolor='black')

    #Configurar etiquetas y diseño
    plt.title(titulo, fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel('Frecuencia / Conteo', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt

#Identificar automáticamente las columnas numéricas del DataFrame
columnas_numericas = forbes_df.select_dtypes(include=['number']).columns

#Iterar sobre cada columna numérica para generar un boxplot individual
for col in columnas_numericas:
    plt.figure(figsize=(8, 4))

    #Crear el boxplot (vert=False lo hace horizontal para que sea más fácil leer etiquetas si aplica)
    forbes_df[col].plot(kind='box', vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue', color='black'))

    #Configurar etiquetas y diseño
    plt.title(f'Diagrama de Caja (Boxplot) de {col}', fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt

# Definir los dos pares específicos solicitados
pares = [
    ('Revenue (Billions)', 'Profits (Billions)'),
    ('Profits (Billions)', 'Market Value (Billions)')
]

for col_x, col_y in pares:
    if col_x in forbes_df.columns and col_y in forbes_df.columns:
        plt.figure(figsize=(9, 5))
        plt.scatter(
            forbes_df[col_x],
            forbes_df[col_y],
            color='teal',
            alpha=0.6,
            edgecolors='k',
            s=40
        )

        plt.title(f'Relación: {col_x} vs {col_y}', fontsize=14)
        plt.xlabel(col_x, fontsize=12)
        plt.ylabel(col_y, fontsize=12)
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.tight_layout()
        plt.show()
    else:
        print(f" No se encontró la columna '{col_x}' o '{col_y}' en el DataFrame.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

#Crear una copia del DataFrame para no alterar los datos originales
df_encoded = forbes_df.copy()

#Convertir todas las columnas categóricas (texto) a códigos numéricos
for col in df_encoded.select_dtypes(include=['object', 'category']).columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

#Calcular la matriz de correlación con TODAS las columnas del dataset
matriz_corr_total = df_encoded.corr()

#Configurar y generar el mapa de calor completo
plt.figure(figsize=(12, 10))
sns.heatmap(
    matriz_corr_total,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1, vmax=1
)


plt.title('Mapa de Calor de Correlación (Todas las Columnas: Numéricas y Categóricas)', fontsize=14, pad=15)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
